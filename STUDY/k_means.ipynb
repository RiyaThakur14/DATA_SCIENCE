```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
df=pd.read_csv("customer_spendings.csv")
df.head()
X=df[["Annual_Income_Thousand","Spending_Score"]]
kmeans=KMeans(n_clusters=4,random_state=42)
df["Cluster"]=kmeans.fit_predict(X)
print(df)
plt.scatter(df["Annual_Income_Thousand"],df["Spending_Score"],c=df["Cluster"])
plt.xlabel("Annual Income")
plt.ylabel("Spending")
plt.title("Customer Groups-KMeans Clustering")
plt.show()
```
